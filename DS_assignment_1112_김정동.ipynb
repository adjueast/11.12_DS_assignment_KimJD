## 1-2. 문자열을 숫자로 변환하기

**요구사항**

① 빈 리스트 `houses`를 생성합니다.

② 반복문을 사용하여 각 행을 순회합니다.

③ 각 행의 값을 아래 자료형으로 변환하여 `houses`에 저장합니다.

| 열 이름 | 변환할 자료형 |
| --- | --- |
| `ID` | `int` |
| `size` | `int` |
| `year` | `int` |
| `price` | `float` |

④ 필수 값이 누락되었거나 숫자로 변환할 수 없는 행은 `houses`에 저장하지 않고, 행 위치와 제외 사유를 출력합니다.

⑤ 한 행에서 변환 오류가 발생해도 반복문을 중단하지 않고 다음 행을 계속 처리한 뒤, 정상 저장 행 수와 제외 행 수를 출력합니다.

---

**✅ 확인 포인트**

- `houses`는 빈 리스트로 생성되어야 합니다.
- 각 행은 반복문으로 처리되어야 합니다.
- `ID`, `size`, `year`는 `int`로 변환되어야 합니다.
- `price`는 `float`로 변환되어야 합니다.
- 누락·변환 오류 행의 위치와 제외 사유, 정상 저장 행 수와 제외 행 수가 출력되어야 합니다.

````python
# ① 빈 리스트 houses를 생성
houses = []

# ② 정상 저장 행수와 제외 행수를 count하기 위한 변수 설정
success = 0
fail = 0

# ③ 반복문을 사용해 각 행을 순회
for i in range(len(df_house)):
  try:
    # ④ 각 행의 자료값을 변환
    id_val = int(df_house.iloc[i,0])
    size_val = int(df_house.iloc[i,1])
    year_val = int(df_house.iloc[i,2])
    price_val = float(df_house.iloc[i,3])

    # ⑤ row_types 딕셔너리에 각 행의 자료 유형 입력
    row_types = {
        "ID" : id_val,
        "size" : size_val,
        "year" : year_val,
        "price" : price_val
    }

    # ⑥ houses 리스트에 각 행의 자료 값 입력
    houses.append(row_types)
    success += 1

  # ⑦ 변환 오류 발생 시 제외 행, 제외 사유 출력
  except (ValueError, TypeError) as e:
    fail += 1
    print(f"{i}번째 행의 제외 사유는 {e}입니다.")

# ⑧ 정상 저장 행수, 제외 행수 출력
print(f"정상 저장 행 수는 {success}개, 제외 행 수는 {fail}개입니다.")
````

② 주택의 `year` 값을 중복 없이 저장하고 출력합니다.

③ 주택 ID로 가격을 조회할 수 있도록 저장하고, ID 3의 가격을 출력합니다.

④ 최소 크기 `30`과 최대 가격 `4`를 **각각 하나의 원소로 갖는 묶음**으로 저장하세요. 생성한 뒤 원소를 추가/삭제하거나 다른 값으로 교체할 수 없는 자료구조를 선택하고, 선택 이유를 설명하세요.

⑤ 각 작업에 리스트·튜플·딕셔너리·집합 중 적합한 자료구조를 선택하고, 선택 이유 및 선택하지 않은 자료구조 1개와의 차이를 코드 주석으로 작성합니다. 순서 유지, 중복 허용 여부, 키를 통한 조회, 변경 가능 여부 중 해당 작업의 판단 근거를 포함합니다.

---

**✅ 확인 포인트**

- ID 목록, 중복 없는 `year`, ID 3의 가격, 불변 검색 조건이 각각 출력되어야 합니다.
- 각 자료구조 선택의 근거와 대안과의 차이가 코드 주석에 포함되어야 합니다.

````python
# =========================================================================
# [문제 1-5] 각 목적에 적합한 자료구조 선택 및 구현
# =========================================================================

# -------------------------------------------------------------------------
# ① houses를 활용하여 전체 주택 ID를 원래 순서대로 저장하고 출력합니다.
# -------------------------------------------------------------------------
# - 선택한 자료구조: 리스트 (List)
# - 선택 이유: 주택 ID들을 원래 데이터 순서 그대로 '순서 유지'하며 다루기에 가장 적합하고, 가변(Mutable) 객체이므로 유연하게 다룰 수 있습니다.
# - 선택하지 않은 자료구조와의 차이 (집합 vs 리스트): 집합(Set)은 순서가 없기 때문에 원래의 순서를 보장할 수 없으며, 중복을 허용하지 않으므로 본 작업에 부적합합니다.
# -------------------------------------------------------------------------
id_list = [house['ID'] for house in houses]
print("① [리스트] 전체 주택 ID 목록 (원래 순서):", id_list)

# -------------------------------------------------------------------------
# ② 주택의 year 값을 중복 없이 저장하고 출력합니다.
# -------------------------------------------------------------------------
# - 선택한 자료구조: 집합 (Set)
# - 선택 이유: 집합(Set)은 내부적으로 '중복 허용 여부'에서 중복을 자동으로 제거해 주므로, 고유한 연식(year) 값들만 추출하기에 가장 직관적이고 효율적입니다.
# - 선택하지 않은 자료구조와의 차이 (리스트 vs 집합): 리스트(List)는 중복 값을 제한 없이 허용하므로, 중복을 제거하려면 별도로 멤버십 검사(if not in)나 복잡한 조건 처리를 거쳐야 하는 번거로움이 있습니다.
# -------------------------------------------------------------------------
year_set = {house['year'] for house in houses}
print("② [집합] 중복 없는 주택 연식(year) 목록:", year_set)

# -------------------------------------------------------------------------
# ③ 주택 ID로 가격을 조회할 수 있도록 저장하고, ID 3의 가격을 출력합니다.
# -------------------------------------------------------------------------
# - 선택한 자료구조: 딕셔너리 (Dictionary)
# - 선택 이유: 주택 ID를 고유 키(Key)로 하고 가격을 값(Value)으로 매핑하여 '키를 통한 빠른 조회'가 가능합니다. 전체 데이터를 반복할 필요 없이 O(1) 시간에 ID로 가격 조회가 가능하여 가장 효율적입니다.
# - 선택하지 않은 자료구조와의 차이 (리스트 vs 딕셔너리): 리스트(List)는 키 조회를 지원하지 않아 특정 ID의 값을 찾으려면 전체 데이터를 순회(Loop)하며 조건문으로 검색해야 하는 성능적 한계가 있습니다.
# -------------------------------------------------------------------------
id_price_dict = {house['ID']: house['price'] for house in houses}
print("③ [딕셔너리] ID 3의 주택 가격 조회:", id_price_dict.get(3))

# -------------------------------------------------------------------------
# ④ 최소 크기 30과 최대 가격 4를 각각 하나의 원소로 갖는 묶음으로 저장하고 설명하기
# -------------------------------------------------------------------------
# - 선택한 자료구조: 튜플 (Tuple)
# - 선택 이유: 튜플은 한 번 생성한 뒤에는 원소를 추가/삭제하거나 다른 값으로 교체할 수 없는 '변경 가능 여부(Immutability, 불변성)'를 특징으로 가집니다. 필터링 기준(최소 크기, 최대 가격)과 같이 오염되지 않고 안전하게 유지되어야 하는 상수 데이터를 저장하기에 완벽한 자료구조입니다.
# - 선택하지 않은 자료구조와의 차이 (리스트 vs 튜플): 리스트(List)는 가변(Mutable) 객체이므로 실행 도중 코드 실수로 인해 내부 기준 원소 값이 다른 값으로 변경되거나 추가/삭제될 우려가 있습니다.
# -------------------------------------------------------------------------
limit_conditions = (30, 4) # (최소 크기, 최대 가격)
print("④ [튜플] 수정 불가능한 검색 조건 묶음:", limit_conditions)
````
